```julia
using Distributions
using GaussianProcesses
using LinearAlgebra
using JLD2
using Plots;
using CairoMakie; 
using TuePlots
T = Theme(
    TuePlots.SETTINGS[:ProbNum25];
    font = true,
    fontsize = true,
    figsize = true,
    single_column = false,
    thinned = true
)
```

```julia
# Time parameters
Δt = 0.01
len_time = 100
tspan = (0.0, Δt*(len_time-1))
tsteps = range(0, step=Δt, length=len_time)

num_samples = 3
```

### Matérn 1/2

```julia
kernel = Mat12Iso(0.0, 0.0)
kmean  = MeanZero()
gp = GP([0.0], [0.0], kmean, kernel)

μ, σ2 = predict_y(gp,tsteps)

paths = zeros(len_time, num_samples)
for tk in 1:len_time
    paths[tk,:] = rand(Normal(μ[tk], sqrt(σ2[tk])), num_samples)
end
```

```julia
with_theme(T) do
    fig = Figure()
    ax = Axis(fig[1, 1], xlabel = "time (sec)", ylabel = "f(t)")

    sp = series!(ax, tsteps, paths')
    # axislegend(ax)
    CairoMakie.ylims!(ax,-3,3)
    hidespines!(ax, :t, :r)
    save("figures/example-nu12.eps", fig, pt_per_unit = 1) # pt_per_unit is needed to ensure the correct sizes
end
```
